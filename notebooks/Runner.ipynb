```python
from pathlib import Path

from ploomber import DAG
from ploomber.executors import Parallel
from ploomber.products import File
from ploomber.tasks import NotebookRunner
from sklearn.model_selection import ParameterGrid
```

```python
MAX_PROCESSES = 4
NOTEBOOK_NAME = "./notebooks/LightGCOT_swiss_roll.ipynb"
```

```python
param_grid = {
    "EPSILON": [0.01],
    "N_POTENTIALS": [50],
    "M_POTENTIALS": [10],
    "EXP_COST": ["MLP"],
    "M_X_UNPAIRED_SAMPLES": [0, 256, 1024],
    "N_Y_UNPAIRED_SAMPLES": [0, 256, 1024],
    "L_PAIRED_SAMPLES": [16],
}
```

```python
dag = DAG(executor=Parallel(processes=MAX_PROCESSES))
```

```python
for params in ParameterGrid(param_grid):
    name = (
        "_".join(f"{key}={val}" for (key, val) in params.items() if key != "EXP_COST")
        + f"EXP_COST={params['EXP_COST']}"
    )
    print(name)
```

```python
for params in ParameterGrid(param_grid):
    if (
        (params["M_X_UNPAIRED_SAMPLES"] == 0 and params["N_Y_UNPAIRED_SAMPLES"] == 0)
        or (params["M_X_UNPAIRED_SAMPLES"] == 256 and params["N_Y_UNPAIRED_SAMPLES"] == 0)
        or (params["M_X_UNPAIRED_SAMPLES"] == 0 and params["N_Y_UNPAIRED_SAMPLES"] == 256)
        or (params["M_X_UNPAIRED_SAMPLES"] == 256 and params["N_Y_UNPAIRED_SAMPLES"] == 256)
    ):
        continue
    name = (
        "_".join(f"{key}={val}" for (key, val) in params.items() if key != "EXP_COST")
        + f"EXP_COST={params['EXP_COST']}"
    )
    NotebookRunner(
        Path(NOTEBOOK_NAME),
        File(f"./ploomber_notebooks/{name}.ipynb"),
        dag=dag,
        local_execution=True,
        name=name,
        params=params,
    )
```

```python
dag.build()
```
